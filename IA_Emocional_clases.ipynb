```python
def asistente_bienestar(mensaje):
    mensaje = mensaje.lower()

    if "estres" in mensaje:
        return "Respira profundo 🙏. Intenta tomar un descanso corto."
    elif "triste" in mensaje:
        return "No estás solo 💙. Hablar ayuda mucho."
    elif "cansado" in mensaje:
        return "Tu cuerpo necesita pausa ⚡. Descansa un poco."
    else:
        return "Estoy aquí para ayudarte 😊"

# Uso
entrada = input("¿Cómo te sientes?: ")
respuesta = asistente_bienestar(entrada)
print(respuesta)
```

```python
# AI Empathy Assistant - Industria 4.0 + 5.0

class MotorReglas:
    def __init__(self):
        self.reglas = {
            "estres": "Respira profundo 🙏. Intenta una pausa de 10 min.",
            "triste": "No estás solo 💙. Hablar con alguien ayuda.",
            "cansado": "Tu cuerpo necesita descanso ⚡.",
            "ansioso": "Intenta la técnica de respiración 4-7-8."
        }

    def analizar(self, mensaje):
        for palabra in self.reglas:
            if palabra in mensaje.lower():
                return palabra
        return "neutral"

    def responder(self, emocion):
        return self.reglas.get(emocion, "Estoy aquí para escucharte 😊")


class AsistenteIA:
    def __init__(self):
        self.motor = MotorReglas()

    def procesar(self, mensaje):
        emocion = self.motor.analizar(mensaje)
        respuesta = self.motor.responder(emocion)
        return emocion, respuesta


# Simulación de interacción
asistente = AsistenteIA()

mensaje = input("¿Cómo te sientes hoy?: ")
emocion, respuesta = asistente.procesar(mensaje)

print("\n🧠 Emoción detectada:", emocion)
print("💬 Respuesta:", respuesta)
```
